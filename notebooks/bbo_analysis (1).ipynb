# BBO Capstone Analysis

This notebook provides a simple summary of query history and returned outputs from the Black-Box Optimisation (BBO) capstone project.

## Purpose
- organise query history by round and function
- track best and latest outputs
- examine changes across rounds
- support transparent and reproducible analysis of the optimisation process


```python
import pandas as pd
```

## Example query history

The table below shows the structure used to record query points, function identifiers, and returned outputs across rounds.


```python
data = [
    {"round": 1, "function": 1, "query": "0.100000-0.200000", "output": 0.12},
    {"round": 2, "function": 1, "query": "0.120000-0.250000", "output": 0.18},
    {"round": 3, "function": 1, "query": "0.130000-0.240000", "output": 0.16},
    {"round": 1, "function": 2, "query": "0.500000-0.400000", "output": -0.32},
    {"round": 2, "function": 2, "query": "0.550000-0.420000", "output": -0.20},
    {"round": 3, "function": 2, "query": "0.600000-0.450000", "output": -0.15},
]

df = pd.DataFrame(data)
df
```

```python
summary = df.groupby("function").agg(
    rounds_recorded=("round", "count"),
    best_output=("output", "max"),
    latest_output=("output", "last"),
)
summary
```

```python
df["change_from_previous"] = df.groupby("function")["output"].diff()
df

```

```python
best_queries = df.loc[df.groupby("function")["output"].idxmax()].sort_values("function")
best_queries
```

## Round-by-round changes

The next table shows how each function's output changed from one round to the next. This helps identify whether a function is improving steadily, plateauing, or behaving inconsistently.


```python
df["change_from_previous"] = df.groupby("function")["output"].diff()
df
```

## Interpretation

This notebook is intended as a lightweight analysis tool rather than a full optimiser. It supports the capstone by making query history easier to inspect and compare across rounds.

Key ideas:
- repeated local gains may justify exploitation
- inconsistent results may justify exploration
- tracking changes over time supports better evidence-based query decisions
